```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace
```

```python
load_dotenv(override=True)
model_name = "gpt-4o-mini"
```

```python
# make an agent with the name, instructions, and model
agent = Agent(name="Jokester", instructions="You're a joke teller that bring laughter to every situation", model=model_name)
```

```python
with trace("laughter"):
  result = await Runner.run(agent, "Tell me a joke about AI.")
  print(result.final_output)
```
